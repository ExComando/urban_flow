### Normalizar fechas:

````python
# Normalizar las de fechas con el formato 'YYYY-MM-DD':
df['fecha'] = pd.to_datetime(df['fecha'], errors='coerce', format='mixed')
df['fecha'].dt.strftime('%Y-%m-%d')
# Completar las fechas inválidas con "1932-01-01":
df['fecha'] = df['fecha'].fillna(pd.to_datetime('1932-01-01'))

print("Primeras 10 fechas normalizadas:")
df['fecha'].head(10)
````

### Normalizar horas:

````python
# Normalizar los horas al formato de 24hs:
df['hora'] = pd.to_datetime(df['hora'], errors='coerce').dt.strftime('%H:%M')
# Las horas inválidas se procesaran como 00:00.
df['hora'] = df['hora'].fillna('00:00')

print("Primeras 10 horas normalizadas:")
df['hora'].head(10)
````

### Normalizar ubicaciones

````python
# Mediante la expresión regular [^a-zA-Z0-9\s] conservo sólo caracteres
# alfanuméricos y espacios (\s).
df['ubicacion'] = df['ubicacion'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)
df['ubicacion'].str.upper()

print("Primeras 10 ubicaciones normalizadas:")
df['ubicacion'].head(10)
````

### Limpiar y normalizar patentes:

````python
 #Quitar caracteres especiales:
df['patente'] = df['patente'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)
# Completo los vacíos con pd.NA
df['patente'] = df['patente'].replace('', pd.NA)
# Completo los erróneos con pd.NA
df['patente'] = df['patente'].fillna(pd.NA)

df['patente'].str.upper()

print("Últimas 10 patentes normalizadas:")
df['patente'].tail(10)
````

### Limpieza de valores relevantes para multas:

````python
# Eliminar las observaciones con valores nulos en los atributos
# "velocidad_registrada", "velocidad_maxima", "radar_id":
atributos_relevantes = ["velocidad_registrada", "velocidad_maxima", "radar_id"]
total_observaciones = df.shape[0]

for atributo in atributos_relevantes:
    df = df.dropna(subset=[atributo])

total_observaciones = total_observaciones - df.shape[0]

print(f"Se eliminaron {total_observaciones} filas.")

print("Las columnas observadas son:")
for atr in atributos_relevantes:
    print(f" - {atr}")
````

### Detección y limpieza de outliers:

````python
# Detectar y eliminar los outliers. Mostrar la cantidad de registros
# eliminados y cuales son las columnas que se observaron

atributos_relevantes = ["velocidad_registrada", "velocidad_maxima"]
total_observaciones = len(df)

#Método del rango intercuartílico (IQR):
serie_vr = pd.to_numeric(df["velocidad_registrada"], errors='coerce')
Q1_vr = serie_vr.quantile(0.25)
Q3_vr = serie_vr.quantile(0.75)
IQR_vr = Q3_vr - Q1_vr
limite_inferior_vr = Q1_vr - 1.5 * IQR_vr
limite_superior_vr = Q3_vr + 1.5 * IQR_vr

#Eliminar outliers según atributo "velocidad_registrada":
df.drop(df[(df["velocidad_registrada"] < limite_inferior_vr) |
      (df["velocidad_registrada"] > limite_superior_vr)].index, inplace=True)

serie_vm = pd.to_numeric(df["velocidad_maxima"], errors='coerce')
Q1_vm = serie_vm.quantile(0.25)
Q3_vm = serie_vm.quantile(0.75)
IQR_vm = Q3_vm - Q1_vm
limite_inferior_vm = Q1_vm - 1.5 * IQR_vm
limite_superior_vm = Q3_vm + 1.5 * IQR_vm

# Eliminar outliers según "velocidad_maxima":
df.drop(df[(df["velocidad_maxima"] < limite_inferior_vm) |
 (df["velocidad_maxima"] > limite_superior_vm)].index, inplace=True)

total_observaciones = total_observaciones - len(df)

print(f"Se eliminaron {total_observaciones} filas.")
print("Las columnas observadas son:")
for atributo in atributos_relevantes:
    print(f" - {atributo}")

if not total_observaciones:
    print("No se encontraron outliers.")
````

### Nuevo atributo `exceso_velocidad_real`:

````python
df["exceso_velocidad_real"] = df["velocidad_registrada"] - df["velocidad_maxima"]

print("Primeras 10 patentes y su exceso de velocidad:")
print(df[["patente", "exceso_velocidad_real"]].head(10).to_string(index=False))
print()
````

### Nuevo atributo `exceso_velocidad`:

````python
df["exceso_velocidad"] = df["velocidad_registrada"] - (df["velocidad_maxima"] * 1.05)
print(df[["patente", "exceso_velocidad"]].head(10).to_string(index=False))
````

### Limpieza de no infractores:

````python
total_observaciones = len(df)
df.drop(df[df["exceso_velocidad"] <= 0].index, inplace=True)
total_observaciones = total_observaciones - len(df)

print(f"Se eliminaron {total_observaciones} filas.")
````
